```python
from pathlib import Path
import pickle
from enterprise.pulsar import Pulsar

# ---- CONFIG ----
TIMING_PKG = "tempo2"   # or "pint"
BASE = Path("/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/dataset_2")
PAR_DIR = BASE / "par"
TIM_DIR = BASE / "tim"
OUT_FILE = BASE / "dataset2_all_pulsars.pkl"

# ---- BUILD ALL ----
pars = sorted(PAR_DIR.glob("*.par"))
pulsars = []
skipped = []

for par in pars:
    tim = TIM_DIR / (par.stem + ".tim")
    if not tim.exists():
        skipped.append((par.name, "missing tim"))
        continue
    try:
        psr = Pulsar(str(par), str(tim), timing_package=TIMING_PKG)
        pulsars.append(psr)
        print(f"✔ {par.stem}: {len(psr.toas)} TOAs")
    except Exception as e:
        skipped.append((par.name, str(e)))

# ---- SAVE SINGLE PKL ----
with open(OUT_FILE, "wb") as f:
    pickle.dump(pulsars, f, protocol=pickle.HIGHEST_PROTOCOL)

print(f"\nSaved {len(pulsars)} pulsars -> {OUT_FILE}")
if skipped:
    print("Skipped:")
    for name, reason in skipped:
        print("  ✖", name, "->", reason)
```

```python
from pathlib import Path
import re, pickle

BASE = Path("/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/dataset_2")
PAR_IN  = BASE / "par_pint_clean"        # from your previous cleaning step
PAR_OUT = BASE / "par_pint_compat"       # new, more compatible copies
TIM_DIR = BASE / "tim"
OUT_PKL = BASE / "dataset2_all_pulsars_pint.pkl"

PAR_OUT.mkdir(exist_ok=True, parents=True)

def decide_binary_model(lines):
    """Infer best replacement model from params present."""
    # Token presence checks (cheap & robust)
    has_eps   = any(re.match(r"\s*(EPS1|EPS2)\b", l) for l in lines)
    has_h3h4  = any(re.match(r"\s*(H3|H4|STIG)\b", l) for l in lines)
    if has_eps and has_h3h4:
        return "ELL1H"
    if has_eps:
        return "ELL1"
    # fall back
    return "DD"

def transform_par_text(text):
    lines = text.splitlines()

    # snapshot presence for model decision
    model_guess = decide_binary_model(lines)

    out = []
    forced_unites = False
    saw_units = False
    saw_timeeph = False
    binary_line_idx = None

    for idx, line in enumerate(lines):
        s = line.strip()

        # keep comments/blank
        if not s or s.startswith("#") or s.startswith("C "):
            out.append(line); continue

        tok = s.split()[0]

        # Force UNITS TDB
        if tok == "UNITS":
            saw_units = True
            if "TDB" not in s:
                out.append("UNITS TDB")
            else:
                out.append(line)
            continue

        # Force TIMEEPH FB90 (drop other settings)
        if tok == "TIMEEPH":
            saw_timeeph = True
            out.append("TIMEEPH FB90")
            continue

        # Drop Tempo2-only / unsupported switches
        if tok == "DILATEFREQ":
            # PINT warns: not supported
            continue

        # Map Tempo2 meta-binary or unsupported models
        if tok == "BINARY":
            model = s.split()[1] if len(s.split()) > 1 else ""
            if model == "T2":
                # Choose specific model based on parameters
                repl = decide_binary_model(lines)
                out.append(f"BINARY {repl}")
                binary_line_idx = len(out)-1
                continue
            if model == "DDH":
                # Prefer ELL1H if EPS* present; else fall back to DD
                repl = "ELL1H" if decide_binary_model(lines) in ("ELL1H","ELL1") else "DD"
                out.append(f"BINARY {repl}")
                binary_line_idx = len(out)-1
                continue

        out.append(line)

    # If no UNITS line, inject TDB at top (some T2 pars omit it)
    if not saw_units:
        out.insert(0, "UNITS TDB")

    # If no TIMEEPH line, inject FB90 (what PINT supports)
    if not saw_timeeph:
        out.insert(1, "TIMEEPH FB90")

    return "\n".join(out) + "\n"

# --- Write transformed PARs ---
transformed = []
for par in sorted(PAR_IN.glob("*.par")):
    raw = par.read_text()
    new = transform_par_text(raw)
    dst = PAR_OUT / par.name
    dst.write_text(new)
    transformed.append(dst)
    print(f"Compat-wrote {par.name}")

# --- Try building pulsars with PINT ---
from enterprise.pulsar import Pulsar
built = []
skipped = []

for par in transformed:
    tim = TIM_DIR / (par.stem + ".tim")
    if not tim.exists():
        skipped.append((par.name, "missing .tim")); continue
    try:
        psr = Pulsar(str(par), str(tim), timing_package="pint")
        built.append(psr)
        print(f"✔ {par.stem}: {len(psr.toas)} TOAs")
    except Exception as e:
        skipped.append((par.name, str(e)))
        print(f"✖ {par.stem} FAILED: {e}")

with open(OUT_PKL, "wb") as f:
    pickle.dump(built, f, protocol=pickle.HIGHEST_PROTOCOL)

print("\nSummary")
print("-------")
print(f"Saved {len(built)} pulsars (PINT) -> {OUT_PKL}")
if skipped:
    print("Remaining skips:", len(skipped))
    for name, reason in skipped:
        print("  ✖", name, "->", reason)
print(f"Transformed PARs are in: {PAR_OUT}")
```
