```python
#!/usr/bin/env python3
import numpy as np
import h5py

# ============================================================
# USER INPUT (change these per file)
# ============================================================
base = "/scratch/na00078/projects/IPTA_MDC2/h5_files/"
filename = "G2D2_broad_detect_outfile.h5"      # <-- Change per run
runtime = 53338.0                              # <-- wall time in seconds

# QuickCW parameter names vs Enterprise parameter names
qcw_param = "0_log10_fgw"
ent_param = "log10_fgw"

# If computing ESS from post–dL-masked QuickCW samples:
use_dL_mask = True
target_d_L = 75.4               # Mpc
d_L_percent_tolerance = 1.0     # ±1% mask
# ============================================================


# ============================================================
# MCMC AUTOCORRELATION AND ESS TOOLS
# ============================================================
def autocorrelation(x):
    """Compute autocorrelation via FFT."""
    x = x - np.mean(x)
    N = len(x)
    fft = np.fft.fft(x, n=2*N)
    ac = np.fft.ifft(fft * np.conjugate(fft))[:N].real
    ac /= ac[0]
    return ac

def integrated_autocorr_time(x, max_lag=5000):
    ac = autocorrelation(x)
    tau = 1.0
    for lag in range(1, min(max_lag, len(ac))):
        if ac[lag] <= 0:
            break
        tau += 2.0 * ac[lag]
    return tau

def compute_ess(x):
    tau = integrated_autocorr_time(x)
    ESS = len(x) / tau
    return ESS, tau


# ============================================================
# LOAD AND PROCESS QUICKCW FILES
# ============================================================
def load_quickcw_samples(f, param_name, apply_mask=True):
    params = [p.decode("utf-8") for p in f["par_names"][()]]
    if param_name not in params:
        raise ValueError(f"{param_name} not in QCW par_names.")

    idx = params.index(param_name)

    # prefer masked samples if present
    if "samples_masked" in f.keys():
        samples_all = f["samples_masked"][:, :]
    else:
        samples_all = f["samples_cold"][0, :, :]

    if not apply_mask:
        return samples_all[:, idx]

    # ========================================================
    # Compute d_L mask
    # ========================================================
    # Extract h, f, Mc
    h_amp  = 10**samples_all[:, params.index("0_log10_h")]
    f_gw   = 10**samples_all[:, params.index("0_log10_fgw")]
    Mchirp = 10**samples_all[:, params.index("0_log10_mc")]

    # Constants
    c = 299792458.0
    megaparsec = 3.086e22
    T_sun = 1.327124400e20 / c**3

    # Compute luminosity distance for each sample
    log10_dL = np.log10(2 * (Mchirp*T_sun)**(5/3) * (np.pi*f_gw)**(2/3) / h_amp * c/megaparsec)
    dL = 10**log10_dL

    # Construct mask
    dL_min = target_d_L * (1 - d_L_percent_tolerance/100)
    dL_max = target_d_L * (1 + d_L_percent_tolerance/100)
    mask = np.where((dL >= dL_min) & (dL <= dL_max))[0]

    if len(mask) == 0:
        raise RuntimeError("dL mask produced zero samples!")

    return samples_all[mask, idx]


# ============================================================
# LOAD ENTERPRISE FILES
# ============================================================
def load_enterprise_samples(f, param_name):
    params = [p.decode("utf-8") for p in f["params"][()]]
    if param_name not in params:
        raise ValueError(f"{param_name} not in Enterprise params.")

    idx = params.index(param_name)

    burn = f["metadata/burn"][()]
    chain = f["chain"][burn:, :]

    return chain[:, idx]


# ============================================================
# MAIN SCRIPT LOGIC
# ============================================================

with h5py.File(base + filename, "r") as f:
    keys = list(f.keys())
    print("Detected keys:", keys)

    # --------------------------
    # QuickCW case
    # --------------------------
    if "par_names" in keys and ("samples_masked" in keys or "samples_cold" in keys):
        print("Detected: QuickCW file")

        samples = load_quickcw_samples(
            f,
            param_name=qcw_param,
            apply_mask=use_dL_mask
        )
        filetype = "QuickCW"

    # --------------------------
    # Enterprise case
    # --------------------------
    elif "chain" in keys and "params" in keys:
        print("Detected: Enterprise file")

        samples = load_enterprise_samples(
            f,
            param_name=ent_param
        )
        filetype = "Enterprise"

    else:
        raise RuntimeError("Unrecognized file type.")


# ============================================================
# COMPUTE ESS & PRINT RESULTS
# ============================================================
ESS, tau = compute_ess(samples)
ESS_sec = ESS / runtime

print("\n================= ESS RESULTS =================")
print("File:", filename)
print("Framework:", filetype)
print("Parameter:", qcw_param if filetype == "QuickCW" else ent_param)
print("Samples used:", len(samples))
print("Autocorrelation time τ =", tau)
print("ESS =", ESS)
print("Runtime =", runtime)
print("ESS/sec =", ESS_sec)
print("================================================\n")
```
