```python
import numpy as np
import pickle
from enterprise.pulsar import Pulsar
from enterprise_extensions import deterministic
from QuickCW import const_mcmc as cm

# ============= USER INPUTS ============= #
pkl_file = "/scratch/na00078/projects/IPTA_MDC2/IPTA_MDC2_data/psr_objects/G2D2_IPTA_MDC2_all_pulsars.pkl"     # path to your simulated pulsar pickle file
pulsar_index = 0                  # pick any pulsar to test
# ---- Injected CW parameters (your actual ones) ---- #
cos_gwtheta = np.cos(0.6387905062299246)
gwphi = 3.3335788713091694
cos_inc = 0.8412486994612669
log10_fgw = np.log10(3.7e-09)
fgw = 10**log10_fgw
log10_mc = np.log10(4.3e9)
log10_h = -13.668773493298787
phase0_inj = 0.24434609527920614
psi_inj = 1.1187560505283651
tref_inj_mjd = 55443.93364609394
# ======================================= #

# 1. Load pulsars
with open(pkl_file, "rb") as f:
    psrs = pickle.load(f)
psr = psrs[pulsar_index]
print("Loaded pulsar:", psr.name)

# 2. Match QuickCW tref to the injection
cm.tref = int(round(tref_inj_mjd * 86400))
print(f"QuickCW tref set to {cm.tref/86400.0:.4f} MJD")

# 3. Build Enterprise deterministic CW signal using the same parameters
cw_func = deterministic.cw_delay(
    cos_gwtheta=cos_gwtheta,
    gwphi=gwphi,
    log10_mc=log10_mc,
    log10_h=log10_h,
    log10_fgw=log10_fgw,
    phase0=phase0_inj,
    psi=psi_inj,
    cos_inc=cos_inc,
    psrTerm=True,
    evolve=True,
    tref=cm.tref,
)

# Evaluate signal at each TOA
res_qcw = cw_func(psr)

# 4. Get the injected (libstempo) residuals from the pulsar object
res_libstempo = psr.residuals  # seconds

# 5. Compare both signals
corr = np.corrcoef(res_libstempo, res_qcw)[0, 1]
if corr > 0.9:
    sign = "SAME"
elif corr < -0.9:
    sign = "OPPOSITE"
else:
    sign = "DIFFERENT"

print(f"Correlation = {corr:.5f}  →  ψ convention: {sign}")

# 6. Interpretation
if sign == "SAME":
    print("✅ libstempo and QuickCW use the same ψ convention. No correction needed.")
elif sign == "OPPOSITE":
    print("⚠️ ψ differs by about π/2 (90°). Subtract π/2 from QuickCW ψ to match libstempo.")
else:
    print("❌ Signals differ by more than a ψ rotation. Check tref, fgw, or other parameters.")
```
