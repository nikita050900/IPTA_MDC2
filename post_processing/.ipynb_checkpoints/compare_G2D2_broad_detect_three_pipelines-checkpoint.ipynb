```python
# ============================================================
# G2D2 broad fGW Detection: 2x2 comparison of three pipelines
# - My QCW subsampled to 75k for visual fairness
# - Loki and Enterprise use full chains
# - SD BF computed from FULL chains
#
# 6x6 corner: cos_inc, log10_fgw, log10_h0, log10_mc, Phi0, psi
#
# Pipelines:
#   - My QCW (dL-masked):  G2D2_broad_detect_dLmasked_75.400Mpc.h5
#   - Loki QCW:            G2D2_broad_detect_loki_75k_04_May_2026.h5
#   - Bjorn Enterprise:    D2_det_varyfgw_core_single.h5
#
# For broad fGW, log10_h0 is derived from sampled log10_fgw, log10_mc,
# and log10_dL for both Loki and Enterprise. Enterprise still has dL
# fixed at injection, but fGW is sampled.
# ============================================================

%matplotlib inline

import numpy as np
import h5py
import matplotlib.pyplot as plt
import corner
from enterprise_extensions import model_utils

plt.rcParams["figure.dpi"] = 100
plt.rcParams["font.size"] = 14

np.random.seed(42)


# ============================================================
# Constants and paths
# ============================================================
C_LIGHT = 299792458.0
MPC_IN_M = 3.085677581491367e22
T_SUN = 1.327124400e20 / C_LIGHT**3

LOG10_T_SUN = np.log10(T_SUN)
LOG10_PI = np.log10(np.pi)
LOG10_C = np.log10(C_LIGHT)
LOG10_MPC = np.log10(MPC_IN_M)
LOG10_2 = np.log10(2.0)

D_MPC_FIXED = 75.400  # Enterprise has dL fixed even in broad fGW

LOKI_FILE = (
    "/scratch/na00078/projects/IPTA_MDC2/h5_files/"
    "G2D2_broad_detect_loki_75k_04_May_2026.h5"
)
QCW_FILE = (
    "/scratch/na00078/projects/IPTA_MDC2/h5_files/dl_masked/"
    "G2D2_broad_detect_dLmasked_75.400Mpc.h5"
)
ENT_FILE = (
    "/scratch/na00078/projects/IPTA_MDC2/h5_files/dl_masked/"
    "ent_files/D2_det_varyfgw_core_single.h5"
)

N_SUBSAMPLE = 75000


# ============================================================
# Truth, labels, ranges
# ============================================================
truth = {
    "0_cos_inc":     0.8412486994612669,
    "0_log10_fgw":   np.log10(3.7e-9),
    "0_log10_h":     -13.668773493298787,
    "0_log10_mc":    np.log10(4.3e9),
    "0_phase0":      0.24434609527920614,
    "0_psi":         1.1187560505283651,
}

qc_keys = [
    "0_cos_inc",
    "0_log10_fgw",
    "0_log10_h",
    "0_log10_mc",
    "0_phase0",
    "0_psi",
]

labels = [
    r"$\cos\iota$",
    r"$\log_{10} f_{\rm GW}$",
    r"$\log_{10} h_0$",
    r"$\log_{10}\mathcal{M}_c$",
    r"$\Phi_0$",
    r"$\psi$",
]

ranges = [
    (-1, 1),
    (-9, -7),
    (-15, -12),
    (8.5, 10.0),
    (0, 2*np.pi),
    (0, np.pi),
]


# ============================================================
# Log-space h0 derivation
# ============================================================
def derive_log10_h0(log10_mc, log10_fgw, log10_dL_Mpc):
    return (
        LOG10_2
        + (5.0/3.0) * (log10_mc + LOG10_T_SUN)
        + (2.0/3.0) * (LOG10_PI + log10_fgw)
        - (log10_dL_Mpc + LOG10_MPC - LOG10_C)
    )


# ============================================================
# Load Loki
# ============================================================
print("Loading Loki samples...")
with h5py.File(LOKI_FILE, "r") as f:
    loki_samples = f["samples_cold"][0, :, :]
    loki_par_names = [p.decode() if isinstance(p, bytes) else p
                      for p in f["par_names"][:]]

loki_log10_mc = loki_samples[:, loki_par_names.index("0_log10_mc")]
loki_log10_fgw = loki_samples[:, loki_par_names.index("0_log10_fgw")]
loki_log10_dist = loki_samples[:, loki_par_names.index("0_log10_dist")]
loki_log10_h0 = derive_log10_h0(loki_log10_mc, loki_log10_fgw, loki_log10_dist)

loki_plot_full = np.column_stack([
    loki_samples[:, loki_par_names.index("0_cos_inc")],
    loki_log10_fgw,
    loki_log10_h0,
    loki_log10_mc,
    loki_samples[:, loki_par_names.index("0_phase0")],
    loki_samples[:, loki_par_names.index("0_psi")],
])
print(f"  Loki: {loki_plot_full.shape}")


# ============================================================
# Load my QCW (full + subsampled)
# ============================================================
print("Loading my QCW dL-masked samples...")
with h5py.File(QCW_FILE, "r") as f:
    qc_samples = f["samples_masked"][...]
    qc_par_names = [p.decode() if isinstance(p, bytes) else p
                    for p in f["par_names"][:]]

qc_plot_full = np.column_stack([qc_samples[:, qc_par_names.index(k)] for k in qc_keys])
print(f"  My QCW (full): {qc_plot_full.shape}")

sub_idx = np.random.choice(qc_plot_full.shape[0], size=N_SUBSAMPLE, replace=False)
qc_plot = qc_plot_full[sub_idx]
print(f"  My QCW (subsampled): {qc_plot.shape}")


# ============================================================
# Load Enterprise (broad fGW, dL fixed at injection)
# ============================================================
print("Loading Enterprise samples...")
with h5py.File(ENT_FILE, "r") as f:
    e_samples_all = f["chain"][...]
    e_par_names = [p.decode() if isinstance(p, bytes) else p
                   for p in f["params"][:]]
    burn = f["metadata/burn"][()]
e_samples = e_samples_all[burn:, :]

ent_log10_mc = e_samples[:, e_par_names.index("log10_mc")]
ent_log10_fgw = e_samples[:, e_par_names.index("log10_fgw")]
ent_log10_h0 = derive_log10_h0(
    ent_log10_mc,
    ent_log10_fgw,                                    # sampled in broad fGW
    np.full_like(ent_log10_mc, np.log10(D_MPC_FIXED)),  # dL still fixed
)

ent_plot = np.column_stack([
    e_samples[:, e_par_names.index("cos_inc")],
    ent_log10_fgw,
    ent_log10_h0,
    ent_log10_mc,
    e_samples[:, e_par_names.index("phase0")],
    e_samples[:, e_par_names.index("psi")],
])
print(f"  Enterprise: {ent_plot.shape}\n")


# ============================================================
# Plot styling
# ============================================================
C_QCW   = "#0072B2"
C_LOKI  = "#009E73"
C_ENT   = "#D55E00"
C_TRUTH = "#CC79A7"


def corner_kwargs(color):
    return dict(
        color=color,
        plot_datapoints=False,
        plot_density=False,
        fill_contours=True,
        smooth=1.0,
        smooth1d=1.0,
        levels=(0.393, 0.865, 0.989),
        hist_kwargs=dict(linewidth=2.0, color=color),
        contour_kwargs=dict(linewidths=1.5, colors=[color]),
        contourf_kwargs=dict(alpha=0.35, colors=[color]),
    )


def make_panel_corner(samples_list, colors_list, labels_pipelines, panel_title):
    truth_vals = [truth[k] for k in qc_keys]

    fig = corner.corner(
        samples_list[0],
        labels=labels,
        truths=truth_vals,
        truth_color=C_TRUTH,
        range=ranges,
        show_titles=False,
        label_kwargs={"fontsize": 14},
        **corner_kwargs(colors_list[0]),
    )

    for samples, color in zip(samples_list[1:], colors_list[1:]):
        corner.corner(
            samples,
            fig=fig,
            labels=labels,
            truths=truth_vals,
            truth_color=C_TRUTH,
            range=ranges,
            show_titles=False,
            **corner_kwargs(color),
        )

    for ax in fig.get_axes():
        ax.tick_params(axis="both", labelsize=12)

    handles = [plt.Line2D([0], [0], color=c, lw=2.5, label=lab)
               for c, lab in zip(colors_list, labels_pipelines)]
    handles.append(plt.Line2D([0], [0], color=C_TRUTH, lw=1.5, label="Injected"))
    fig.legend(handles=handles, loc="upper right", fontsize=12,
               bbox_to_anchor=(0.98, 0.98), frameon=True)

    fig.suptitle(panel_title, fontsize=16, y=1.0)
    return fig


# ============================================================
# Generate four panels
# ============================================================
print("Panel 1: My QCW vs Loki")
fig1 = make_panel_corner(
    [qc_plot, loki_plot_full],
    [C_QCW, C_LOKI],
    ["My QCW (dL-masked, 75k)", "Loki QCW (direct dL, 75k)"],
    "G2D2 broad fGW: My QCW vs Loki QCW",
)
plt.show()

print("Panel 2: Loki vs Enterprise")
fig2 = make_panel_corner(
    [loki_plot_full, ent_plot],
    [C_LOKI, C_ENT],
    ["Loki QCW (direct dL, 75k)", "Enterprise (75k)"],
    "G2D2 broad fGW: Loki QCW vs Enterprise",
)
plt.show()

print("Panel 3: My QCW vs Enterprise")
fig3 = make_panel_corner(
    [qc_plot, ent_plot],
    [C_QCW, C_ENT],
    ["My QCW (dL-masked, 75k)", "Enterprise (75k)"],
    "G2D2 broad fGW: My QCW vs Enterprise",
)
plt.show()

print("Panel 4: All three pipelines")
fig4 = make_panel_corner(
    [qc_plot, loki_plot_full, ent_plot],
    [C_QCW, C_LOKI, C_ENT],
    ["My QCW (dL-masked, 75k)", "Loki QCW (direct dL, 75k)", "Enterprise (75k)"],
    "G2D2 broad fGW: All three pipelines",
)
plt.show()


# ============================================================
# SD Bayes factors using FULL chains
# ============================================================
print("\n" + "="*60)
print("Savage-Dickey Bayes factors (FULL chains, no subsampling)")
print("="*60)

qcw_BF, qcw_BF_err = model_utils.bayes_fac(
    samples=qc_plot_full[:, qc_keys.index("0_log10_h")], logAmax=-11)
print(f"My QCW (n={qc_plot_full.shape[0]:,}):     B10 = {qcw_BF:.4f} +/- {qcw_BF_err:.4f}")

loki_BF, loki_BF_err = model_utils.bayes_fac(
    samples=loki_plot_full[:, qc_keys.index("0_log10_h")], logAmax=-11)
print(f"Loki   (n={loki_plot_full.shape[0]:,}):     B10 = {loki_BF:.4f} +/- {loki_BF_err:.4f}")
print("  Note: Loki log10_h0 derived; implicit prior on h0 is non-log-uniform.")

ent_BF, ent_BF_err = model_utils.bayes_fac(
    samples=ent_plot[:, qc_keys.index("0_log10_h")], logAmax=-11)
print(f"Ent    (n={ent_plot.shape[0]:,}):     B10 = {ent_BF:.4f} +/- {ent_BF_err:.4f}")
print("  Note: Enterprise log10_h0 derived with sampled fGW and fixed dL.")
```

```python
qcw_log10_h = qc_plot_full[:, qc_keys.index("0_log10_h")]
loki_log10_h = loki_plot_full[:, qc_keys.index("0_log10_h")]
ent_log10_h = ent_plot[:, qc_keys.index("0_log10_h")]

print(f"QCW max log10_h0:  {qcw_log10_h.max():.3f}")
print(f"Loki max log10_h0: {loki_log10_h.max():.3f}")
print(f"Ent max log10_h0:  {ent_log10_h.max():.3f}")

print(f"QCW samples within 0.5 of -11: {(qcw_log10_h > -11.5).sum()}")
print(f"Loki samples within 0.5 of -11: {(loki_log10_h > -11.5).sum()}")
print(f"Ent samples within 0.5 of -11: {(ent_log10_h > -11.5).sum()}")
```

```python
# ============================================================
# Diagnostic: SD BF tail check at log10_h0 = -18 boundary
# ============================================================

# Critical threshold: SD BF uses bins between logAmin and logAmin + 0.1
# So samples with log10_h0 < -17.9 are what populate the BF estimator
THRESHOLD = -17.9

print("="*60)
print(f"Samples below log10_h0 = {THRESHOLD} (the SD BF window)")
print("="*60)

for name, chain in [("My QCW", qcw_log10_h), 
                    ("Loki", loki_log10_h), 
                    ("Enterprise", ent_log10_h)]:
    n_below = (chain < THRESHOLD).sum()
    frac = n_below / len(chain)
    print(f"  {name:12s}: {n_below:6d} / {len(chain):6d} samples ({100*frac:.3f}%), "
          f"min log10_h0 = {chain.min():.3f}")

# Bin-by-bin within the SD BF range
print("\n" + "="*60)
print("Sample counts per delta bin (logAmin = -18)")
print("(bayes_fac requires > 200 samples per bin for that bin to count)")
print("="*60)

for delta in [0.01, 0.02, 0.05, 0.1]:
    upper = -18 + delta
    print(f"\n  Bin: log10_h0 in [-18, {upper:.2f}]")
    for name, chain in [("My QCW", qcw_log10_h),
                        ("Loki", loki_log10_h),
                        ("Enterprise", ent_log10_h)]:
        n_in_bin = ((chain >= -18) & (chain <= upper)).sum()
        marker = " <-- BF valid" if n_in_bin > 200 else ""
        print(f"    {name:12s}: {n_in_bin:6d} samples{marker}")
```

```python
samples = ent_log10_h
logAmin, logAmax = -18, -11   # default
prior = 1 / (logAmax - logAmin)  # = 0.25
N = len(samples)
dA = np.linspace(0.01, 0.1, 100)

bf_values = []
for delta in dA:
    n = np.sum(samples <= (logAmin + delta))
    if n > 200:
        post = n / N / delta
        bf_values.append(prior / post)

print(f"Number of valid bins: {len(bf_values)}")
print(f"Mean BF: {np.mean(bf_values):.4f}")
print(f"Min BF in valid bins: {min(bf_values):.4f}")
print(f"Max BF in valid bins: {max(bf_values):.4f}")
```

```python
# ============================================================
# Diagnostic: Enterprise broad fGW chain trace plots
# Check for mode switching behavior
# ============================================================

%matplotlib inline
import numpy as np
import h5py
import matplotlib.pyplot as plt

plt.rcParams["figure.dpi"] = 100
plt.rcParams["font.size"] = 11


# ============================================================
# Reload Enterprise with full chain (before burn-in)
# ============================================================
ENT_FILE = (
    "/scratch/na00078/projects/IPTA_MDC2/h5_files/dl_masked/"
    "ent_files/D2_det_varyfgw_core_single.h5"
)

with h5py.File(ENT_FILE, "r") as f:
    e_full = f["chain"][...]
    e_par_names = [p.decode() if isinstance(p, bytes) else p
                   for p in f["params"][:]]
    burn = f["metadata/burn"][()]

print(f"Total samples (pre burn-in): {e_full.shape[0]}")
print(f"Burn-in: {burn}")
print(f"Post burn-in samples: {e_full.shape[0] - burn}")


# ============================================================
# Pull out CW + GWB params for trace plots
# ============================================================
ent_log10_mc_full = e_full[:, e_par_names.index("log10_mc")]
ent_log10_fgw_full = e_full[:, e_par_names.index("log10_fgw")]
ent_log10_h_full = e_full[:, e_par_names.index("log10_h0")] if "log10_h0" in e_par_names else None
ent_cos_inc_full = e_full[:, e_par_names.index("cos_inc")]

# Check if GWB amplitude is in the chain
gwb_idx = None
for name in e_par_names:
    if "gwb" in name.lower() and "log10_a" in name.lower():
        gwb_idx = e_par_names.index(name)
        print(f"Found GWB amplitude: {name}")
        break

ent_gwb_log10_A = e_full[:, gwb_idx] if gwb_idx is not None else None


# ============================================================
# Trace plots: full chain (pre + post burn-in)
# ============================================================
n_panels = 4 if ent_gwb_log10_A is not None else 3
fig, axes = plt.subplots(n_panels, 1, figsize=(14, 3*n_panels), sharex=True)

axes[0].plot(ent_log10_mc_full, lw=0.3, color="steelblue")
axes[0].axhline(np.log10(4.3e9), color="magenta", lw=1.5, label="injected")
axes[0].axvline(burn, color="red", lw=1, ls="--", label="burn-in")
axes[0].set_ylabel(r"$\log_{10}\mathcal{M}_c$")
axes[0].set_title("Enterprise broad fGW: full chain trace")
axes[0].legend(loc="upper right")

axes[1].plot(ent_log10_fgw_full, lw=0.3, color="steelblue")
axes[1].axhline(np.log10(3.7e-9), color="magenta", lw=1.5, label="injected")
axes[1].axvline(burn, color="red", lw=1, ls="--")
axes[1].set_ylabel(r"$\log_{10}f_{\rm GW}$")
axes[1].legend(loc="upper right")

if ent_log10_h_full is not None:
    axes[2].plot(ent_log10_h_full, lw=0.3, color="steelblue")
    axes[2].axhline(-13.668773, color="magenta", lw=1.5, label="injected")
    axes[2].axvline(burn, color="red", lw=1, ls="--")
    axes[2].set_ylabel(r"$\log_{10}h_0$")
    axes[2].legend(loc="upper right")
else:
    axes[2].plot(ent_cos_inc_full, lw=0.3, color="steelblue")
    axes[2].axhline(0.8412, color="magenta", lw=1.5, label="injected")
    axes[2].axvline(burn, color="red", lw=1, ls="--")
    axes[2].set_ylabel(r"$\cos\iota$")
    axes[2].legend(loc="upper right")

if ent_gwb_log10_A is not None:
    axes[3].plot(ent_gwb_log10_A, lw=0.3, color="steelblue")
    axes[3].axvline(burn, color="red", lw=1, ls="--")
    axes[3].set_ylabel(r"GWB $\log_{10}A$")
    axes[3].set_xlabel("sample index")

axes[-1].set_xlabel("sample index")
plt.tight_layout()
plt.show()


# ============================================================
# Zoom into post burn-in only
# ============================================================
ent_post = e_full[burn:, :]
ent_log10_mc = ent_post[:, e_par_names.index("log10_mc")]
ent_log10_fgw = ent_post[:, e_par_names.index("log10_fgw")]
ent_log10_h_post = ent_log10_h_full[burn:] if ent_log10_h_full is not None else None
ent_gwb_post = ent_gwb_log10_A[burn:] if ent_gwb_log10_A is not None else None

n_panels = 4 if ent_gwb_post is not None else 3
fig, axes = plt.subplots(n_panels, 1, figsize=(14, 3*n_panels), sharex=True)

axes[0].plot(ent_log10_mc, lw=0.3, color="darkorange")
axes[0].axhline(np.log10(4.3e9), color="magenta", lw=1.5, label="injected")
axes[0].set_ylabel(r"$\log_{10}\mathcal{M}_c$")
axes[0].set_title("Enterprise broad fGW: post burn-in only")
axes[0].legend(loc="upper right")

axes[1].plot(ent_log10_fgw, lw=0.3, color="darkorange")
axes[1].axhline(np.log10(3.7e-9), color="magenta", lw=1.5, label="injected")
axes[1].set_ylabel(r"$\log_{10}f_{\rm GW}$")
axes[1].legend(loc="upper right")

if ent_log10_h_post is not None:
    axes[2].plot(ent_log10_h_post, lw=0.3, color="darkorange")
    axes[2].axhline(-13.668773, color="magenta", lw=1.5, label="injected")
    axes[2].axhline(-17.9, color="red", lw=1, ls="--", label="SD BF window")
    axes[2].set_ylabel(r"$\log_{10}h_0$")
    axes[2].legend(loc="upper right")

if ent_gwb_post is not None:
    axes[3].plot(ent_gwb_post, lw=0.3, color="darkorange")
    axes[3].set_ylabel(r"GWB $\log_{10}A$")
    axes[3].set_xlabel("sample index (post burn-in)")

axes[-1].set_xlabel("sample index (post burn-in)")
plt.tight_layout()
plt.show()


# ============================================================
# Highlight low-h0 segments (the noise floor visits)
# ============================================================
if ent_log10_h_post is not None:
    LOW_THRESHOLD = -17.9
    low_mask = ent_log10_h_post < LOW_THRESHOLD
    n_low = low_mask.sum()
    
    print(f"\nSamples with log10_h0 < {LOW_THRESHOLD}: {n_low} / {len(ent_log10_h_post)}")
    
    if n_low > 0:
        # Where do these low samples sit in the chain?
        low_indices = np.where(low_mask)[0]
        print(f"\nFirst 10 indices of low-h0 samples: {low_indices[:10]}")
        print(f"Last 10 indices of low-h0 samples: {low_indices[-10:]}")
        
        # Are they clustered or spread out?
        # If mode switching, expect contiguous chunks
        # If just bad MCMC noise, expect random scatter
        gaps = np.diff(low_indices)
        print(f"\nGaps between consecutive low-h0 samples:")
        print(f"  Median gap: {np.median(gaps):.0f}")
        print(f"  Max gap: {gaps.max()}")
        print(f"  Min gap: {gaps.min()}")
        print(f"  Number of gaps == 1 (consecutive): {(gaps == 1).sum()}")
        
        # Visualize: highlight low-h0 regions on the trace
        fig, ax = plt.subplots(figsize=(14, 4))
        ax.plot(ent_log10_h_post, lw=0.3, color="lightgray", label="all samples")
        ax.scatter(low_indices, ent_log10_h_post[low_mask], 
                   s=4, color="red", label=f"log10_h0 < {LOW_THRESHOLD}", zorder=5)
        ax.axhline(-13.668773, color="magenta", lw=1.5, label="injected")
        ax.axhline(LOW_THRESHOLD, color="red", lw=1, ls="--", alpha=0.5)
        ax.set_xlabel("sample index (post burn-in)")
        ax.set_ylabel(r"$\log_{10}h_0$")
        ax.set_title("Enterprise broad fGW: noise-floor visits highlighted")
        ax.legend(loc="lower right")
        plt.tight_layout()
        plt.show()
```

```python
# ============================================================
# Trace plot comparison across all three pipelines
# Mode switching diagnostic for G2D2 broad fGW
# ============================================================

%matplotlib inline
import numpy as np
import h5py
import matplotlib.pyplot as plt

plt.rcParams["figure.dpi"] = 100
plt.rcParams["font.size"] = 11


# ============================================================
# Constants and paths (broad fGW)
# ============================================================
C_LIGHT = 299792458.0
MPC_IN_M = 3.085677581491367e22
T_SUN = 1.327124400e20 / C_LIGHT**3
LOG10_T_SUN = np.log10(T_SUN)
LOG10_PI = np.log10(np.pi)
LOG10_C = np.log10(C_LIGHT)
LOG10_MPC = np.log10(MPC_IN_M)
LOG10_2 = np.log10(2.0)

D_MPC_FIXED = 75.400

LOKI_FILE = (
    "/scratch/na00078/projects/IPTA_MDC2/h5_files/"
    "G2D2_broad_detect_loki_75k_04_May_2026.h5"
)
QCW_FILE = (
    "/scratch/na00078/projects/IPTA_MDC2/h5_files/dl_masked/"
    "G2D2_broad_detect_dLmasked_75.400Mpc.h5"
)
ENT_FILE = (
    "/scratch/na00078/projects/IPTA_MDC2/h5_files/dl_masked/"
    "ent_files/D2_det_varyfgw_core_single.h5"
)


def derive_log10_h0(log10_mc, log10_fgw, log10_dL_Mpc):
    return (LOG10_2
            + (5.0/3.0) * (log10_mc + LOG10_T_SUN)
            + (2.0/3.0) * (LOG10_PI + log10_fgw)
            - (log10_dL_Mpc + LOG10_MPC - LOG10_C))


# Truth values
truth_mc = np.log10(4.3e9)
truth_fgw = np.log10(3.7e-9)
truth_h0 = -13.668773


# ============================================================
# Load all three chains
# ============================================================
print("Loading my QCW...")
with h5py.File(QCW_FILE, "r") as f:
    qc = f["samples_masked"][...]
    qc_pn = [p.decode() if isinstance(p, bytes) else p for p in f["par_names"][:]]
qcw_mc = qc[:, qc_pn.index("0_log10_mc")]
qcw_fgw = qc[:, qc_pn.index("0_log10_fgw")]
qcw_h = qc[:, qc_pn.index("0_log10_h")]

print("Loading Loki...")
with h5py.File(LOKI_FILE, "r") as f:
    lk = f["samples_cold"][0, :, :]
    lk_pn = [p.decode() if isinstance(p, bytes) else p for p in f["par_names"][:]]
loki_mc = lk[:, lk_pn.index("0_log10_mc")]
loki_fgw = lk[:, lk_pn.index("0_log10_fgw")]
loki_dist = lk[:, lk_pn.index("0_log10_dist")]
loki_h = derive_log10_h0(loki_mc, loki_fgw, loki_dist)

print("Loading Enterprise...")
with h5py.File(ENT_FILE, "r") as f:
    en = f["chain"][...]
    en_pn = [p.decode() if isinstance(p, bytes) else p for p in f["params"][:]]
ent_mc = en[:, en_pn.index("log10_mc")]
ent_fgw = en[:, en_pn.index("log10_fgw")]
ent_h = derive_log10_h0(ent_mc, ent_fgw,
                        np.full_like(ent_mc, np.log10(D_MPC_FIXED)))

print(f"\nChain lengths: My QCW = {len(qcw_mc)}, Loki = {len(loki_mc)}, Ent = {len(ent_mc)}")


# ============================================================
# 3 rows (params) x 3 columns (pipelines) trace plot
# ============================================================
fig, axes = plt.subplots(3, 3, figsize=(18, 10), sharex="col")

C_QCW = "#0072B2"
C_LOKI = "#009E73"
C_ENT = "#D55E00"
C_TRUTH = "#CC79A7"

# Row 0: log10_mc
axes[0, 0].plot(qcw_mc, lw=0.3, color=C_QCW)
axes[0, 0].axhline(truth_mc, color=C_TRUTH, lw=1.5, label="injected")
axes[0, 0].set_ylabel(r"$\log_{10}\mathcal{M}_c$")
axes[0, 0].set_title(f"My QCW (n={len(qcw_mc)})")
axes[0, 0].legend(loc="upper right", fontsize=9)

axes[0, 1].plot(loki_mc, lw=0.3, color=C_LOKI)
axes[0, 1].axhline(truth_mc, color=C_TRUTH, lw=1.5)
axes[0, 1].set_title(f"Loki QCW (n={len(loki_mc)})")

axes[0, 2].plot(ent_mc, lw=0.3, color=C_ENT)
axes[0, 2].axhline(truth_mc, color=C_TRUTH, lw=1.5)
axes[0, 2].set_title(f"Enterprise (n={len(ent_mc)})")

# Row 1: log10_fgw
axes[1, 0].plot(qcw_fgw, lw=0.3, color=C_QCW)
axes[1, 0].axhline(truth_fgw, color=C_TRUTH, lw=1.5)
axes[1, 0].set_ylabel(r"$\log_{10}f_{\rm GW}$")

axes[1, 1].plot(loki_fgw, lw=0.3, color=C_LOKI)
axes[1, 1].axhline(truth_fgw, color=C_TRUTH, lw=1.5)

axes[1, 2].plot(ent_fgw, lw=0.3, color=C_ENT)
axes[1, 2].axhline(truth_fgw, color=C_TRUTH, lw=1.5)

# Row 2: log10_h0
axes[2, 0].plot(qcw_h, lw=0.3, color=C_QCW)
axes[2, 0].axhline(truth_h0, color=C_TRUTH, lw=1.5, label="injected")
axes[2, 0].set_ylabel(r"$\log_{10}h_0$")
axes[2, 0].set_xlabel("sample index")
axes[2, 0].legend(loc="lower right", fontsize=9)

axes[2, 1].plot(loki_h, lw=0.3, color=C_LOKI)
axes[2, 1].axhline(truth_h0, color=C_TRUTH, lw=1.5)
axes[2, 1].set_xlabel("sample index")

axes[2, 2].plot(ent_h, lw=0.3, color=C_ENT)
axes[2, 2].axhline(truth_h0, color=C_TRUTH, lw=1.5)
axes[2, 2].set_xlabel("sample index")

# Match y-axis ranges across columns for fair comparison
for row in range(3):
    ymins, ymaxs = [], []
    for col in range(3):
        yl, yh = axes[row, col].get_ylim()
        ymins.append(yl)
        ymaxs.append(yh)
    ymin, ymax = min(ymins), max(ymaxs)
    for col in range(3):
        axes[row, col].set_ylim(ymin, ymax)

plt.suptitle("G2D2 broad fGW: chain trace comparison", fontsize=14, y=1.00)
plt.tight_layout()
plt.show()


# ============================================================
# Summary stats
# ============================================================
print("\n" + "="*70)
print("Mode switching summary")
print("="*70)

for name, mc, fgw, h in [
    ("My QCW", qcw_mc, qcw_fgw, qcw_h),
    ("Loki", loki_mc, loki_fgw, loki_h),
    ("Enterprise", ent_mc, ent_fgw, ent_h),
]:
    # Define "wrong mode" loosely: Mc < 8 OR fGW > -7.5
    wrong_mode = (mc < 8) | (fgw > -7.5)
    h_at_floor = h < -17.9
    print(f"\n{name}:")
    print(f"  Samples in 'wrong mode' (Mc<8 or fGW>-7.5): "
          f"{wrong_mode.sum()} ({100*wrong_mode.mean():.2f}%)")
    print(f"  Samples at h0 floor (log10_h0 < -17.9): "
          f"{h_at_floor.sum()} ({100*h_at_floor.mean():.3f}%)")
```

```python
# ============================================================
# ESS diagnostic: G2D2 broad fGW
# Both Mc AND fGW are shape parameters
# ============================================================

from emcee import autocorr

def compute_ess(samples):
    try:
        tau = autocorr.integrated_time(samples, quiet=True)[0]
        ess = len(samples) / tau
        return tau, ess
    except Exception:
        return np.nan, np.nan


print("="*70)
print("Effective Sample Size: G2D2 broad fGW")
print("="*70)
print(f"{'Pipeline':<15} {'Param':<10} {'N raw':<12} {'tau':<10} {'ESS':<10}")
print("-"*70)

# Use whichever variables are loaded in your broad fGW notebook
# Adjust names if needed
for pipe_name, mc, fgw, h in [
    ("My QCW",     qcw_mc,  qcw_fgw,  qcw_h),
    ("Loki",       loki_mc, loki_fgw, loki_h),
    ("Enterprise", ent_mc,  ent_fgw,  ent_h),
]:
    tau_mc,  ess_mc  = compute_ess(mc)
    tau_fgw, ess_fgw = compute_ess(fgw)
    tau_h,   ess_h   = compute_ess(h)
    print(f"{pipe_name:<15} {'log10_Mc':<10} {len(mc):<12} {tau_mc:<10.1f} {ess_mc:<10.0f}")
    print(f"{pipe_name:<15} {'log10_fGW':<10} {len(fgw):<12} {tau_fgw:<10.1f} {ess_fgw:<10.0f}")
    print(f"{pipe_name:<15} {'log10_h0':<10} {len(h):<12} {tau_h:<10.1f} {ess_h:<10.0f}")
    print()

print("="*70)
print("Interpretation")
print("="*70)
print("Shape parameter ESS is what matters for Loki's concern.")
print("If QCW Mc/fGW ESS is comparable to Enterprise, the comparison is fair.")
```
