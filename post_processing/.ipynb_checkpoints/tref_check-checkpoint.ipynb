```python
import h5py

# ------------------------------------------------------------
# INPUT FILE
# ------------------------------------------------------------
# COMMENT IF OUTFILE GENERATED
# Thinning script for one source
i = "/scratch/na00078/projects/IPTA_MDC2/h5_files/G2D1_broad_detect_1e9_fixed_gamma_outfile.h5"
infile = i
first_n_param = 8

print(infile)
print(first_n_param)

# ------------------------------------------------------------
# READ SAMPLES + METADATA
# ------------------------------------------------------------
with h5py.File(infile, 'r') as f:

    # T ladder
    Ts = f['T-ladder'][...]

    # Cold samples (only first_n_param)
    samples_cold = f['samples_cold'][:, :, :first_n_param]
    print(samples_cold[-1].shape)

    # Log-likelihood (cold chain only)
    log_likelihood = f['log_likelihood'][:1, :]
    print(log_likelihood.shape)

    # Parameter names
    par_names = [x.decode('UTF-8') for x in list(f['par_names'])]

    # Diagnostics
    acc_fraction = f['acc_fraction'][...]
    fisher_diag = f['fisher_diag'][...]

    # --------------------------------------------------------
    # NEW: CHECK tref USED IN THIS QuickCW RUN
    # --------------------------------------------------------
    tref = None

    # Case 1: QuickCW >=2023 stores tref in pta.attrs
    if "pta" in f and "tref" in f["pta"].attrs:
        tref = f["pta"].attrs["tref"]

    # Case 2: Some QuickCW versions nest pta/pta
    elif "pta" in f and "pta" in f["pta"] and "tref" in f["pta"]["pta"].attrs:
        tref = f["pta"]["pta"].attrs["tref"]

    # Print result
    if tref is not None:
        print("tref found:", tref)
    else:
        print("tref not stored in this HDF5 file")

```
