```python
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt
import corner
import h5py
import scipy.constants as sc
import re
from pathlib import Path

# ---------- SETTINGS ----------
target_d_L = 75.4  # Mpc
base_dir = Path("/scratch/na00078/projects/IPTA_MDC2/h5_files/")
base_name = "IPTA_MDC2_G2D2_detection_"

# Input files
files = [
    "G2D2_narrow_detect_simulated_psi_1p8850_outfile.h5",
    "G2D2_narrow_detect_simulated_psi_2p5133_outfile.h5",
    "G2D2_narrow_detect_simulated_psi_1p2566_outfile.h5",
    "G2D2_narrow_detect_simulated_psi_3p1416_outfile.h5",
    "G2D2_narrow_detect_simulated_psi_0p00pi_outfile.h5",
    "G2D2_narrow_detect_simulated_psi_0p6283_outfile.h5",
]

# ---------- TRUE PARAMS ----------
xxx0 = {
    "0_cos_gwtheta": np.cos(0.6387905062299246),
    "0_cos_inc": 0.8412486994612669,
    "0_gwphi": 3.3335788713091694,
    "0_log10_fgw": np.log10(3.7e-9),
    "0_log10_h": -13.668773493298787,
    "0_log10_mc": np.log10(4.3e9),
    "phase0": 0.24434609527920614,
    "psi": None,  # set dynamically per file
    "distance": 75.4,
}

labels = [
    r"$\cos \iota$",
    r"$\log_{10} A_{\rm e}$",
    r"$\log_{10} {\cal M}$",
    r"$\Phi_0$",
    r"$\psi$"
]
label_to_key = {
    r"$\cos \iota$": "0_cos_inc",
    r"$\log_{10} A_{\rm e}$": "0_log10_h",
    r"$\log_{10} {\cal M}$": "0_log10_mc",
    r"$\Phi_0$": "phase0",
    r"$\psi$": "psi"
}

ranges = [(-1,1), (-18,-11), (9,10), (0,2*np.pi), (0,np.pi)]

# ---------- CONSTANTS ----------
megaparsec = 3.086e+22  # meter
speed_of_light = 299792458.0  # m/s
T_sun = 1.327124400e20 / speed_of_light**3  # G*M_sun/c^3 in seconds

# ---------- LOOP THROUGH FILES ----------
for fname in files:
    infile = base_dir / fname
    print(f"\nProcessing {infile.name}")

   
    # --- extract ψ value from filename (robust) ---
    psi_match = re.search(r"psi_([0-9p]+)(pi)?", fname)
    if psi_match:
        num_part = psi_match.group(1).replace("p", ".").strip(".")
        has_pi = psi_match.group(2) is not None
        try:
            psi_val = float(num_part)
        except ValueError:
            psi_val = 0.0
        if has_pi:
            psi_val *= np.pi
    else:
        psi_val = np.nan
        
    xxx0["psi"] = psi_val
    truths = [xxx0[label_to_key[l]] for l in labels]

    # --- load samples ---
    with h5py.File(infile, "r") as f:
        samples_cold = f["samples_cold"][:, :, :8]
        par_names = [x.decode("UTF-8") for x in list(f["par_names"])]

    # --- reconstruct distance mask ---
    h_amp = 10 ** samples_cold[0][:, 4]
    fff = 10 ** samples_cold[0][:, 3]
    mmm = 10 ** samples_cold[0][:, 5]
    log10_d_L = np.log10(
        2 * (mmm * T_sun) ** (5/3) * (np.pi * fff) ** (2/3)
        / h_amp * speed_of_light / megaparsec
    )

    d_L_min = target_d_L * 0.99
    d_L_max = target_d_L * 1.01
    d_L_mask = np.where((10**log10_d_L >= d_L_min) & (10**log10_d_L <= d_L_max))[0]

    samples2plot = np.concatenate(
        (samples_cold[0][:, :8], np.array([log10_d_L]).T), axis=1
    )

    samples2plot_detect = np.vstack((
        samples2plot[d_L_mask,1],
        samples2plot[d_L_mask,4],
        samples2plot[d_L_mask,5],
        samples2plot[d_L_mask,6],
        samples2plot[d_L_mask,7]
    )).T

    # ---------- CORNER PLOT ----------
    fig = corner.corner(
        samples2plot_detect,
        labels=labels,
        show_titles=True,
        range=ranges,
        hist_kwargs={"density": True},
        truths=truths,
        truth_color="red"
    )

    # beautify
    for ax in fig.get_axes():
        ax.tick_params(axis="both", labelsize=14)
        ax.xaxis.label.set_size(16)
        ax.yaxis.label.set_size(16)

    # prior lines
    for i, ax in enumerate(fig.axes):
        if i == 0:  # cos iota
            Xs = np.linspace(-1, 1)
            ax.plot(Xs, Xs*0+0.5, color="xkcd:green")
        elif i == (len(labels)+1):  # log10 A_e
            Xs = np.linspace(-18, -11)
            ax.plot(Xs, Xs*0+1/7, color="xkcd:green")
        elif i == 2*(len(labels)+1):  # log10 Mc
            Xs = np.linspace(9, 10)
            ax.plot(Xs, Xs*0+1/3, color="xkcd:green")
        elif i == 3*(len(labels)+1):  # phase0
            Xs = np.linspace(0, 2*np.pi)
            ax.plot(Xs, Xs*0+1/(2*np.pi), color="xkcd:green")
        elif i == 4*(len(labels)+1):  # psi
            Xs = np.linspace(0, np.pi)
            ax.plot(Xs, Xs*0+1/np.pi, color="xkcd:green")

    # ---------- TITLE ----------
    psi_over_pi = psi_val / np.pi
    fig.suptitle(
        rf"{base_name} Corner Plot ($\psi_\mathrm{{inj}} = {psi_over_pi:.2f}\,\pi\,\mathrm{{rad}}$)",
        fontsize=22, y=1.05
    )

    plt.tight_layout()
    plt.show()
```
