```python
import pickle
import json
import numpy as np

# -------------------------------------------------
# helper: theta → declination
# Enterprise uses: theta = colatitude = pi/2 - dec
# -------------------------------------------------
def theta_to_dec(theta):
    return np.pi/2 - theta   # radians

# -------------------------------------------------
# load PKL
# -------------------------------------------------
pkl_file = "/scratch/na00078/projects/IPTA_MDC2/IPTA_MDC2_data/psr_objects/G2D2_IPTA_MDC2_all_pulsars.pkl"

with open(pkl_file, "rb") as f:
    pulsars = pickle.load(f)   # list of Tempo2Pulsar objects

# -------------------------------------------------
# build output structure
# format:
# { "J1713+0747": [ [x,y,z], [RA, Dec] ], ... }
# -------------------------------------------------
out = {}

for p in pulsars:

    name = p.name

    # RA and Dec in radians
    ra = float(p.phi)                # RA
    dec = float(theta_to_dec(p.theta))  # Dec

    # Cartesian unit vector
    x, y, z = map(float, p.pos)

    out[name] = [
        [x, y, z],
        [ra, dec]
    ]

# -------------------------------------------------
# save to JSON
# -------------------------------------------------
with open("pulsar_positions.json", "w") as f:
    json.dump(out, f, indent=4)

print("Saved: pulsar_positions.json")
```
