```python
from __future__ import division

import numpy as np
import glob
import matplotlib.pyplot as plt
import scipy.linalg as sl
import os
import json

import enterprise
from enterprise.pulsar import Pulsar
import enterprise.signals.parameter as parameter
from enterprise.signals import utils
from enterprise.signals import signal_base
from enterprise.signals import selections
from enterprise.signals.selections import Selection
from enterprise.signals import white_signals
from enterprise.signals import gp_signals

import corner
from PTMCMCSampler.PTMCMCSampler import PTSampler as ptmcmc

timdir = '/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/dataset_1/'
pardir = '/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/dataset_1/par'
chaindir = '/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/dataset_1/chains/'
figdir = '/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/dataset_1/figs/'
noisedir = '/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/dataset_1/'
noisefile = '/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/group2_psr_noise.json'
#% matplotlib inline
#% config InlineBackend.figure_format = 'retina'
#% load_ext line_profiler
```

```python
parfiles = sorted(glob.glob(pardir + '*.par'))
timfiles = sorted(glob.glob(timdir + '*.tim'))
```

```python
psrs = []
for p, t in zip(parfiles, timfiles):
    psr = Pulsar(p, t)
    psrs.append(psr)

##### parameters and priors #####

# Uniform prior on EFAC, EQUAD
efac = parameter.Uniform(0.1, 7.0)
equad = parameter.Uniform(-9, -5)

# red noise parameters 
# Uniform in log10 Amplitude and in spectral index
log10_A = parameter.Uniform(-18,-12)
gamma = parameter.Uniform(0,5)

##### Set up signals #####

# white noise
ef = white_signals.MeasurementNoise(efac=efac)
eq = white_signals.TNEquadNoise(log10_tnequad=equad)
#eq = white_signals.EquadNoise(log10_equad=equad)

#meas = white_signals.MeasurementNoise(efac=efac, log10_t2equad=equad)


# red noise (powerlaw with 30 frequencies)
pl = utils.powerlaw(log10_A=log10_A, gamma=gamma)
rn = gp_signals.FourierBasisGP(spectrum=pl, components=30)

# timing model
tm = gp_signals.TimingModel()

# full model is sum of components
model = ef + eq + rn + tm
```

```python
psr_dict = {}

for psr in psrs:
    print('Working on ' + psr.name)
    outdir = chaindir + psr.name + '/'
    os.makedirs(outdir, exist_ok=True)

    # initialize PTA
    pta = signal_base.PTA([model(psr)])

    #Pick random initial sampling
    xs = {par.name: par.sample() for par in pta.params}

    # dimension of parameter space
    ndim = len(xs)

    # initial jump covariance matrix
    cov = np.diag(np.ones(ndim) * 0.01**2)

    # Now we figure out which indices the white & red noise parameters have
    wn_idx1 = pta.param_names.index(psr.name + '_efac')
    wn_idx2 = pta.param_names.index(psr.name + '_log10_equad')
    rn_idx1 = pta.param_names.index(psr.name + '_red_noise_log10_A')
    rn_idx2 = pta.param_names.index(psr.name + '_red_noise_gamma')

    # set up jump groups by red noise groups
    ndim = len(xs)
    groups  = [range(0, ndim)]
    #groups.extend([[[rn_idx1,rn_idx2]])
    groups.extend([[wn_idx1,wn_idx2], [rn_idx1,rn_idx2]])

    # intialize sampler
    sampler = ptmcmc(ndim, pta.get_lnlikelihood, pta.get_lnprior, cov, groups=groups, outDir=outdir)

    # sampler for N steps
    N = 100000
    x0 = np.hstack(p.sample() for p in pta.params)
    sampler.sample(x0, N, SCAMweight=30, AMweight=15, DEweight=50)


    chain = np.loadtxt(outdir + 'chain_1.txt')
    pars = sorted(xs.keys())
    burn = int(0.25 * chain.shape[0])

    #Get true noise values for pulsar to plot in corner plot (truth values)
    with open(noisefile, 'r') as nf:
        inputvals = json.load(nf)
        nf.close()
    #Unpacking dictionaries in json file to get at noise values
    noise_vals = inputvals[psr.name]
    truths = [noise_vals["efac"], noise_vals["equad"], noise_vals["rn_spec_ind"], noise_vals["rn_log10_A"]]

    #Make corner plots
    corner.corner(chain[burn:,:-4], 30, truths = truths, labels=pars)
    #corner.corner(chain, 30, truths = truths, labels=pars)
    plt.savefig(figdir + psr.name + '_cornerplt.png')
    plt.close()

    #make dictionary of pulsar parameters from these runs
    psr_dict[psr.name] = {}
    for param_name,idx in zip(pars,range(len(pars))):
        psr_dict[psr.name][param_name] = np.mean(chain[:,idx])
```

```python
# Now we want to save this all as a json file
with open(noisedir+"fit_psr_noise.json", 'w') as fpn:
    json.dump(psr_dict, fpn)
    fpn.close()
```
