```python
import sys
import math 
import numpy as np
import matplotlib.pyplot as plt
import libstempo as T
import glob

T.data = '/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/dataset_1'
```

```python
parfiles = sorted(glob.glob(T.data + '/*.par'))
timfiles = sorted(glob.glob(T.data + '/*.tim'))
```

```python
psrs = []
for p, t in zip(parfiles, timfiles):
    psr = T.tempopulsar(parfile = p, timfile = t)
    psrs.append(psr)
```

```python
import libstempo.plot as LP

for P in psrs:
    P['DM'].fit = False
    P['DM1'].fit = False
    P['DM2'].fit = False
    #P.fit(iters=10)
    P.savepar("./newpars_2b/" + P.name + ".par")
```

```python
import libstempo.plot as LP
newpars = sorted(glob.glob('./newpars_2b/*.par'))
newpsrs = []
for p, t in zip(newpars, timfiles):
    psr = T.tempopulsar(parfile = p, timfile = t)
    newpsrs.append(psr)
for P in newpsrs:
    plt.figure()
    LP.plotres(P,alpha=0.5)
```
