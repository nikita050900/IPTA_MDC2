```python
import numpy as np, libstempo as T
from pathlib import Path
import matplotlib.pyplot as plt, gc

# --- paths ---
BASE = Path("/scratch/na00078/projects/IPTA_MDC2/mdc2/group2/dataset_2")
PAR = BASE / "par/J0030+0451.par"
TIM = BASE / "tim/J0030+0451.tim"

# --- CW params (dataset2) ---
CW = dict(
    fgw=3.7e-9,
    log10_h=-13.668773493298787,
    gw_phi=3.3335788713091694,
    gw_theta=0.6387905062299246,
    inclination=0.8412486994612669,
    phase0=0.24434609527920614,
    psi=1.1187560505283651,
)

# --- derived ---
h0   = 10**CW["log10_h"]
g_ra = CW["gw_phi"]
g_dec= np.pi/2 - CW["gw_theta"]
```

```python
def get_radec(psr):
    """Return RA/DEC in radians, using RAJ/DECJ or ELONG/ELAT."""
    keys = {k.upper(): k for k in psr.pars()}
    if "RAJ" in keys and "DECJ" in keys:
        return float(psr[keys["RAJ"]].val), float(psr[keys["DECJ"]].val)
    if "ELONG" in keys and "ELAT" in keys:
        lon=float(psr[keys["ELONG"]].val); lat=float(psr[keys["ELAT"]].val)
        eps=np.deg2rad(23.43929111)
        sinβ,cosβ=np.sin(lat),np.cos(lat)
        sinλ,cosλ=np.sin(lon),np.cos(lon)
        sinδ=sinβ*np.cos(eps)+cosβ*np.sin(eps)*sinλ
        δ=np.arcsin(np.clip(sinδ,-1,1))
        α=np.arctan2(sinλ*np.cos(eps)-np.tan(lat)*np.sin(eps),cosλ)% (2*np.pi)
        return α,δ
    raise KeyError("no coordinates")

def antenna_factors(p_ra,p_dec,g_ra,g_dec,psi):
    ph=np.array([np.cos(p_dec)*np.cos(p_ra),
                 np.cos(p_dec)*np.sin(p_ra),
                 np.sin(p_dec)])
    kh=np.array([np.cos(g_dec)*np.cos(g_ra),
                 np.cos(g_dec)*np.sin(g_ra),
                 np.sin(g_dec)])
    x=np.array([-np.sin(g_ra),np.cos(g_ra),0.0])
    x-=x.dot(kh)*kh; x/=np.linalg.norm(x)
    y=np.cross(kh,x)
    c2,s2=np.cos(2*psi),np.sin(2*psi)
    eplus=np.outer(x,x)-np.outer(y,y)
    ecross=np.outer(x,y)+np.outer(y,x)
    eplus,ecross=c2*eplus+s2*ecross,-s2*eplus+c2*ecross
    denom=1.0-ph.dot(kh)
    Fp=0.5*(ph@eplus@ph)/denom
    Fx=0.5*(ph@ecross@ph)/denom
    return Fp,Fx

def cw_residuals(psr,use_psrterm=True):
    """Earth + optional pulsar term residuals."""
    pra,pdec=get_radec(psr)
    Fp,Fx=antenna_factors(pra,pdec,g_ra,g_dec,CW["psi"])
    toas=psr.stoas.astype(float)*86400.0
    tref=float(psr["PEPOCH"].val)*86400.0
    cth=np.cos(CW["inclination"])
    fgw=CW["fgw"]; phi0=CW["phase0"]
    phiE=2*np.pi*fgw*(toas-tref)+phi0
    splusE=0.5*(1+cth**2)*np.cos(phiE)
    scrossE=cth*np.sin(phiE)
    if not use_psrterm:
        return (h0/(2*np.pi*fgw))*(Fp*splusE+Fx*scrossE)
    # Pulsar term
    try: px=float(psr["PX"].val); pdist=1.0/px if px>0 else 1.0
    except: pdist=1.0
    tau=pdist*3.085677581e19/299792458.0
    phiP=2*np.pi*fgw*(toas-tref-tau)+phi0
    splusP=0.5*(1+cth**2)*np.cos(phiP)
    scrossP=cth*np.sin(phiP)
    return (h0/(2*np.pi*fgw))*(Fp*(splusE-splusP)+Fx*(scrossE-scrossP))
```

```python
# Load pulsar
psr=T.tempopulsar(str(PAR),str(TIM))
toas=psr.stoas.astype(float)*86400.0

# Residuals
r_norm = cw_residuals(psr,use_psrterm=True)
CW_rot = CW.copy(); CW_rot["psi"]=(CW["psi"]+np.pi/2)%np.pi; CW_rot["phase0"]=(CW["phase0"]+np.pi)%(2*np.pi)
r_rot  = cw_residuals(psr,use_psrterm=True)

# Phase comparison
plt.figure(figsize=(8,4))
plt.plot(toas/86400.0, r_norm*1e6, label='normal')
plt.plot(toas/86400.0, r_rot*1e6,  label='rotated', alpha=0.7)
plt.xlabel("MJD"); plt.ylabel("Residual [μs]")
plt.legend(); plt.tight_layout(); plt.show()

# Measure relative phase shift
from scipy.signal import correlate
corr=np.correlate(r_norm-np.mean(r_norm), r_rot-np.mean(r_rot), mode='full')
lag=np.arange(-len(r_norm)+1,len(r_norm))
lag_s=lag[np.argmax(corr)]/(1/CW["fgw"])/len(r_norm)
print(f"Approx phase shift ≈ {lag_s*2*np.pi:.2f} rad")
gc.collect()
```
